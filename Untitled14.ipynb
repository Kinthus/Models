```python
def find_image_dataset(root_directory):
    """
    Automatically searches for a directory containing:
        train/
        validation/
        test/
    """

    root = Path(root_directory)

    candidates = []

    for directory in root.rglob("*"):

        if not directory.is_dir():
            continue

        required_dirs = [
            directory / "train",
            directory / "validation",
            directory / "test"
        ]

        if all(d.exists() and d.is_dir() for d in required_dirs):
            candidates.append(directory)

    if len(candidates) == 0:
        raise FileNotFoundError(
            "Could not find image_dataset containing train, validation and test."
        )

    # Prefer a directory named image_dataset
    image_dataset_candidates = [
        p for p in candidates if p.name.lower() == "image_dataset"
    ]

    if len(image_dataset_candidates) > 0:
        return image_dataset_candidates[0]

    return candidates[0]


DATASET_ROOT = find_image_dataset(EXTRACT_ROOT)

print("Dataset located at:")
print(DATASET_ROOT)
```

```python
VALID_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff",
    ".webp",
    ".gif"
}


def collect_images(split_path):
    """
    Collect image paths from Fake and Real directories.
    """

    records = []

    for class_name in CLASS_NAMES:

        class_dir = split_path / class_name

        for path in class_dir.rglob("*"):

            if path.is_file() and path.suffix.lower() in VALID_EXTENSIONS:

                records.append({
                    "path": str(path),
                    "label": CLASS_TO_IDX[class_name],
                    "class_name": class_name
                })

    return pd.DataFrame(records)


train_df = collect_images(DATASET_ROOT / "train")

validation_df = collect_images(DATASET_ROOT / "validation")

test_df = collect_images(DATASET_ROOT / "test")


print("Train images:", len(train_df))
print("Validation images:", len(validation_df))
print("Test images:", len(test_df))

print("\nTrain:")
print(train_df["class_name"].value_counts())

print("\nValidation:")
print(validation_df["class_name"].value_counts())

print("\nTest:")
print(test_df["class_name"].value_counts())
```

```python
def validate_images(df, split_name):
    """
    Validate every image using PIL.

    Returns:
        valid_df
        corrupted_df
    """

    valid_records = []
    corrupted_records = []

    print("\n" + "=" * 70)
    print(f"CHECKING {split_name.upper()} IMAGES")
    print("=" * 70)

    for _, row in tqdm(
        df.iterrows(),
        total=len(df),
        desc=f"Checking {split_name}"
    ):

        image_path = row["path"]

        try:

            # First verify image integrity
            with Image.open(image_path) as img:
                img.verify()

            # Open again and force pixel loading
            with Image.open(image_path) as img:
                img = ImageOps.exif_transpose(img)
                img.convert("RGB").load()

            valid_records.append(row.to_dict())

        except Exception as e:

            corrupted_records.append({
                "path": image_path,
                "label": row["label"],
                "class_name": row["class_name"],
                "error": str(e)
            })

    valid_df = pd.DataFrame(valid_records)

    corrupted_df = pd.DataFrame(corrupted_records)

    print("\nValid images:", len(valid_df))
    print("Corrupted/unreadable images:", len(corrupted_df))

    if len(corrupted_df) > 0:

        print("\nCorrupted files:")

        for path in corrupted_df["path"]:
            print(" -", path)

    return valid_df, corrupted_df


train_valid_df, train_corrupt_df = validate_images(
    train_df,
    "train"
)

validation_valid_df, validation_corrupt_df = validate_images(
    validation_df,
    "validation"
)

test_valid_df, test_corrupt_df = validate_images(
    test_df,
    "test"
)


# Combine corruption reports
corrupted_all = pd.concat(
    [
        train_corrupt_df.assign(split="train"),
        validation_corrupt_df.assign(split="validation"),
        test_corrupt_df.assign(split="test")
    ],
    ignore_index=True
)

corrupted_report_path = "/content/corrupted_images_report.csv"

corrupted_all.to_csv(
    corrupted_report_path,
    index=False
)

print("\nCorruption report saved to:")
print(corrupted_report_path)
```

```python
# EfficientNetB0 pretrained on ImageNet expects 224x224 RGB images.
#
# ImageNet normalization:
# mean = [0.485, 0.456, 0.406]
# std  = [0.229, 0.224, 0.225]

IMAGENET_MEAN = [
    0.485,
    0.456,
    0.406
]

IMAGENET_STD = [
    0.229,
    0.224,
    0.225
]


train_transform = transforms.Compose([

    transforms.Resize(
        (256, 256)
    ),

    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.80, 1.0)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2,
        hue=0.05
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        IMAGENET_MEAN,
        IMAGENET_STD
    )
])


eval_transform = transforms.Compose([

    transforms.Resize(
        (256, 256)
    ),

    transforms.CenterCrop(
        IMAGE_SIZE
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        IMAGENET_MEAN,
        IMAGENET_STD
    )
])


print("Transforms created successfully.")
```

```python
class SafeImageDataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):

        return len(self.df)

    def __getitem__(self, index):

        row = self.df.iloc[index]

        image_path = row["path"]
        label = int(row["label"])

        try:

            image = Image.open(
                image_path
            )

            image = ImageOps.exif_transpose(
                image
            )

            image = image.convert(
                "RGB"
            )

            if self.transform is not None:

                image = self.transform(
                    image
                )

            return image, label, image_path

        except Exception as e:

            # This should normally not happen because
            # corrupted files were removed before training.
            #
            # We raise a clear error instead of silently
            # ignoring the problem.

            raise RuntimeError(
                f"Could not read image:\n"
                f"{image_path}\n"
                f"Error: {e}"
            )

```

```python
def create_loader(
    dataframe,
    transform,
    shuffle=False
):

    dataset = SafeImageDataset(
        dataframe,
        transform=transform
    )

    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        drop_last=False
    )

    return loader


print("DataLoader function ready.")
print("NUM_WORKERS =", NUM_WORKERS)
```

```python
def create_model():

    # Load EfficientNetB0 with ImageNet pretrained weights
    weights = models.EfficientNet_B0_Weights.DEFAULT

    model = models.efficientnet_b0(
        weights=weights
    )

    # Freeze the pretrained backbone initially
    for parameter in model.features.parameters():
        parameter.requires_grad = False

    # Replace original ImageNet classification head
    input_features = model.classifier[1].in_features

    model.classifier = nn.Sequential(

        nn.Dropout(
            p=0.30
        ),

        nn.Linear(
            input_features,
            NUM_CLASSES
        )
    )

    # Classification head is trainable
    for parameter in model.classifier.parameters():
        parameter.requires_grad = True

    model = model.to(DEVICE)

    return model


model_test = create_model()


total_parameters = sum(
    p.numel()
    for p in model_test.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in model_test.parameters()
    if p.requires_grad
)

frozen_parameters = (
    total_parameters -
    trainable_parameters
)

total_layers = sum(
    1
    for _ in model_test.modules()
)

trainable_layers = sum(
    1
    for module in model_test.modules()
    if any(
        p.requires_grad
        for p in module.parameters(recurse=False)
    )
)

frozen_layers = total_layers - trainable_layers


print("=" * 70)
print("EFFICIENTNETB0 MODEL INFORMATION")
print("=" * 70)

print("Model: EfficientNetB0")
print("Pretrained weights: ImageNet")
print("Input size:", IMAGE_SIZE, "x", IMAGE_SIZE)
print("Number of classes:", NUM_CLASSES)
print("Classes:", CLASS_NAMES)

print("\nParameters:")
print("Total:", f"{total_parameters:,}")
print("Trainable:", f"{trainable_parameters:,}")
print("Frozen:", f"{frozen_parameters:,}")

print("\nApproximate module counts:")
print("Total modules:", total_layers)
print("Trainable modules:", trainable_layers)
print("Frozen modules:", frozen_layers)


del model_test

if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

```python
def calculate_metrics(
    y_true,
    y_pred,
    y_probability
):

    metrics = {}

    metrics["accuracy"] = accuracy_score(
        y_true,
        y_pred
    )

    metrics["precision"] = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    metrics["recall"] = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    metrics["f1"] = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    if len(np.unique(y_true)) == 2:

        metrics["roc_auc"] = roc_auc_score(
            y_true,
            y_probability
        )

        metrics["average_precision"] = (
            average_precision_score(
                y_true,
                y_probability
            )
        )

    else:

        metrics["roc_auc"] = np.nan
        metrics["average_precision"] = np.nan

    return metrics
```

```python
AMP_ENABLED = torch.cuda.is_available()


def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer
):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels, _ in loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        with torch.amp.autocast(
            device_type=DEVICE.type,
            enabled=AMP_ENABLED
        ):

            outputs = model(images)

            loss = criterion(
                outputs,
                labels
            )

        if AMP_ENABLED:

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

        else:

            loss.backward()

            optimizer.step()

        running_loss += (
            loss.item() *
            images.size(0)
        )

        predictions = outputs.argmax(
            dim=1
        )

        correct += (
            predictions == labels
        ).sum().item()

        total += labels.size(0)

    epoch_loss = running_loss / total

    epoch_accuracy = correct / total

    return epoch_loss, epoch_accuracy


@torch.no_grad()
def evaluate_model(
    model,
    loader,
    criterion
):

    model.eval()

    running_loss = 0.0

    all_labels = []
    all_predictions = []
    all_probabilities = []

    total = 0

    for images, labels, _ in loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )

        with torch.amp.autocast(
            device_type=DEVICE.type,
            enabled=AMP_ENABLED
        ):

            outputs = model(images)

            loss = criterion(
                outputs,
                labels
            )

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        predictions = outputs.argmax(
            dim=1
        )

        running_loss += (
            loss.item() *
            images.size(0)
        )

        total += labels.size(0)

        all_labels.extend(
            labels.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            probabilities[:, 1]
            .cpu()
            .numpy()
        )

    average_loss = running_loss / total

    all_labels = np.array(
        all_labels
    )

    all_predictions = np.array(
        all_predictions
    )

    all_probabilities = np.array(
        all_probabilities
    )

    metrics = calculate_metrics(
        all_labels,
        all_predictions,
        all_probabilities
    )

    return (
        average_loss,
        metrics,
        all_labels,
        all_predictions,
        all_probabilities
    )
```

```python
# ============================================================
# 5-FOLD STRATIFIED CROSS VALIDATION
# ============================================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

X = cv_df["path"].values

y = cv_df["label"].values


fold_results = []

fold_histories = []

oof_labels = []

oof_predictions = []

oof_probabilities = []

fold_best_epochs = []


print("=" * 80)
print("STARTING 5-FOLD STRATIFIED CROSS VALIDATION")
print("=" * 80)

for fold, (train_indices, val_indices) in enumerate(
    skf.split(X, y),
    start=1
):

    print("\n")
    print("=" * 80)
    print(f"FOLD {fold} / 5")
    print("=" * 80)

    fold_train_df = cv_df.iloc[
        train_indices
    ].reset_index(drop=True)

    fold_val_df = cv_df.iloc[
        val_indices
    ].reset_index(drop=True)


    # --------------------------------------------------------
    # Show fold distribution
    # --------------------------------------------------------

    print("\nTraining distribution:")
    print(
        fold_train_df[
            "class_name"
        ].value_counts()
    )

    print("\nValidation distribution:")
    print(
        fold_val_df[
            "class_name"
        ].value_counts()
    )

    print(
        "\nFold training images:",
        len(fold_train_df)
    )

    print(
        "Fold validation images:",
        len(fold_val_df)
    )


    # --------------------------------------------------------
    # DataLoaders
    # --------------------------------------------------------

    train_loader = create_loader(
        fold_train_df,
        train_transform,
        shuffle=True
    )

    val_loader = create_loader(
        fold_val_df,
        eval_transform,
        shuffle=False
    )


    # --------------------------------------------------------
    # Create fresh EfficientNetB0
    # --------------------------------------------------------

    model = create_model()


    # --------------------------------------------------------
    # Loss
    # --------------------------------------------------------

    criterion = nn.CrossEntropyLoss()


    # --------------------------------------------------------
    # Optimizer
    # --------------------------------------------------------

    trainable_params = [
        p for p in model.parameters()
        if p.requires_grad
    ]

    optimizer = torch.optim.AdamW(
        trainable_params,
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )


    # --------------------------------------------------------
    # Learning-rate scheduler
    # --------------------------------------------------------

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=MAX_EPOCHS
    )


    # --------------------------------------------------------
    # Mixed precision scaler
    # --------------------------------------------------------

    if AMP_ENABLED:

        scaler = torch.amp.GradScaler(
            "cuda"
        )

    else:

        scaler = None


    # --------------------------------------------------------
    # History
    # --------------------------------------------------------

    history = {
        "epoch": [],
        "train_loss": [],
        "val_loss": [],
        "train_accuracy": [],
        "val_accuracy": [],
        "precision": [],
        "recall": [],
        "f1": [],
        "roc_auc": [],
        "average_precision": [],
        "learning_rate": []
    }


    best_val_f1 = -np.inf

    best_epoch = 1

    patience_counter = 0

    fold_checkpoint = (
        f"/content/fold_{fold}_best.pth"
    )


    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    for epoch in range(1, MAX_EPOCHS + 1):

        train_loss, train_accuracy = train_one_epoch(
            model,
            train_loader,
            criterion,
            optimizer
        )


        (
            val_loss,
            val_metrics,
            val_labels,
            val_predictions,
            val_probabilities
        ) = evaluate_model(
            model,
            val_loader,
            criterion
        )


        current_lr = optimizer.param_groups[0]["lr"]


        history["epoch"].append(epoch)

        history["train_loss"].append(
            train_loss
        )

        history["val_loss"].append(
            val_loss
        )

        history["train_accuracy"].append(
            train_accuracy
        )

        history["val_accuracy"].append(
            val_metrics["accuracy"]
        )

        history["precision"].append(
            val_metrics["precision"]
        )

        history["recall"].append(
            val_metrics["recall"]
        )

        history["f1"].append(
            val_metrics["f1"]
        )

        history["roc_auc"].append(
            val_metrics["roc_auc"]
        )

        history["average_precision"].append(
            val_metrics["average_precision"]
        )

        history["learning_rate"].append(
            current_lr
        )


        print(
            f"Epoch {epoch:02d}/{MAX_EPOCHS} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Train Acc: {train_accuracy:.4f} | "
            f"Val Acc: {val_metrics['accuracy']:.4f} | "
            f"Precision: {val_metrics['precision']:.4f} | "
            f"Recall: {val_metrics['recall']:.4f} | "
            f"F1: {val_metrics['f1']:.4f} | "
            f"ROC-AUC: {val_metrics['roc_auc']:.4f} | "
            f"AP: {val_metrics['average_precision']:.4f}"
        )


        # ----------------------------------------------------
        # Save best model based on validation F1
        # ----------------------------------------------------

        if val_metrics["f1"] > best_val_f1:

            best_val_f1 = val_metrics["f1"]

            best_epoch = epoch

            patience_counter = 0

            torch.save(
                {
                    "model_state_dict": model.state_dict(),
                    "fold": fold,
                    "epoch": epoch,
                    "best_val_f1": best_val_f1
                },
                fold_checkpoint
            )

        else:

            patience_counter += 1


        scheduler.step()


        # ----------------------------------------------------
        # Early stopping
        # ----------------------------------------------------

        if patience_counter >= PATIENCE:

            print(
                f"\nEarly stopping at epoch {epoch}."
            )

            break


    # --------------------------------------------------------
    # Load best checkpoint
    # --------------------------------------------------------

    checkpoint = torch.load(
        fold_checkpoint,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )


    # --------------------------------------------------------
    # Final evaluation of this fold
    # --------------------------------------------------------

    (
        final_val_loss,
        final_val_metrics,
        final_labels,
        final_predictions,
        final_probabilities
    ) = evaluate_model(
        model,
        val_loader,
        criterion
    )


    fold_best_epochs.append(
        best_epoch
    )


    fold_result = {
        "fold": fold,
        "best_epoch": best_epoch,
        "accuracy": final_val_metrics["accuracy"],
        "precision": final_val_metrics["precision"],
        "recall": final_val_metrics["recall"],
        "f1": final_val_metrics["f1"],
        "roc_auc": final_val_metrics["roc_auc"],
        "average_precision": final_val_metrics["average_precision"]
    }

    fold_results.append(
        fold_result
    )

    fold_histories.append(
        pd.DataFrame(history)
    )


    # --------------------------------------------------------
    # Store out-of-fold predictions
    # --------------------------------------------------------

    oof_labels.extend(
        final_labels.tolist()
    )

    oof_predictions.extend(
        final_predictions.tolist()
    )

    oof_probabilities.extend(
        final_probabilities.tolist()
    )


    print("\nBest epoch:", best_epoch)

    print(
        "Best validation F1:",
        f"{best_val_f1:.4f}"
    )


    del model
    del train_loader
    del val_loader

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


print("\n")
print("=" * 80)
print("5-FOLD CROSS VALIDATION COMPLETED")
print("=" * 80)
```

```python
class EfficientNetGradCAM:

    def __init__(
        self,
        model
    ):

        self.model = model

        self.activations = None

        self.target_layer = None

        # Find the last Conv2d layer
        for module in reversed(
            list(
                model.features.modules()
            )
        ):

            if isinstance(
                module,
                nn.Conv2d
            ):

                self.target_layer = module

                break


        if self.target_layer is None:

            raise RuntimeError(
                "Could not find a Conv2d layer."
            )


        self.forward_handle = (
            self.target_layer.register_forward_hook(
                self._forward_hook
            )
        )


    def _forward_hook(
        self,
        module,
        inputs,
        output
    ):

        self.activations = output


    def generate(
        self,
        image_path,
        target_class=None
    ):

        # Store original requires_grad states for model.features parameters
        original_feature_grad_states = {}
        for param in self.model.features.parameters():
            original_feature_grad_states[param] = param.requires_grad
            param.requires_grad = True # Temporarily enable gradients

        self.model.eval()

        image = Image.open(
            image_path
        )

        image = ImageOps.exif_transpose(
            image
        )

        image = image.convert(
            "RGB"
        )

        input_tensor = eval_transform(
            image
        ).unsqueeze(
            0
        ).to(DEVICE)

        # Ensure the input tensor also requires grad for this pass
        input_tensor.requires_grad_(True)

        self.model.zero_grad(
            set_to_none=True
        )

        # Perform the forward pass within a context where gradients are enabled
        with torch.enable_grad():
            output = self.model(
                input_tensor
            )

        probabilities = torch.softmax(
            output,
            dim=1
        )

        if target_class is None:

            target_class = int(
                torch.argmax(
                    probabilities,
                    dim=1
                ).item()
            )


        score = output[
            0,
            target_class
        ]


        gradients = torch.autograd.grad(
            score,
            self.activations,
            retain_graph=False
        )[0]


        activations = self.activations[
            0
        ]

        gradients = gradients[
            0
        ]


        # Global average pooling of gradients
        weights = gradients.mean(
            dim=(1, 2)
        )


        # Weighted activation maps
        cam = torch.sum(
            weights[:, None, None] *
            activations,
            dim=0
        )


        cam = torch.relu(
            cam
        )


        cam -= cam.min()

        if cam.max() > 0:

            cam /= cam.max()


        cam = cam.detach().cpu().numpy()


        # Resize heatmap
        cam_image = Image.fromarray(
            np.uint8(
                cam * 255
            )
        )

        cam_image = cam_image.resize(
            image.size,
            Image.Resampling.BILINEAR
        )


        # Restore original requires_grad states for model.features parameters
        for param in self.model.features.parameters():
            param.requires_grad = original_feature_grad_states[param]

        # Restore input_tensor state
        input_tensor.requires_grad_(False)

        return (
            image,
            cam_image,
            probabilities[
                0
            ].detach().cpu().numpy(),
            target_class
        )


    def close(self):

        self.forward_handle.remove()


print("Grad-CAM implementation ready.")
```

```python
inference_code = r'''
import json
import argparse

import torch
import torch.nn as nn

from PIL import Image, ImageOps

from torchvision import models, transforms


# ============================================================
# Configuration
# ============================================================

MODEL_PATH = "final_model.pth"

CONFIG_PATH = "model_config.json"

CLASS_NAMES_PATH = "class_names.json"


DEVICE = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)


# ============================================================
# Load configuration
# ============================================================

with open(CONFIG_PATH, "r") as f:
    config = json.load(f)


with open(CLASS_NAMES_PATH, "r") as f:
    class_names = json.load(f)


IMAGE_SIZE = config["image_size"]

MEAN = config["normalization"]["mean"]

STD = config["normalization"]["std"]


# ============================================================
# Image preprocessing
# ============================================================

transform = transforms.Compose([

    transforms.Resize(
        (256, 256)
    ),

    transforms.CenterCrop(
        IMAGE_SIZE
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        MEAN,
        STD
    )
])


# ============================================================
# Create EfficientNetB0
# ============================================================

model = models.efficientnet_b0(
    weights=None
)


input_features = (
    model.classifier[1].in_features
)


model.classifier = nn.Sequential(

    nn.Dropout(
        p=0.30
    ),

    nn.Linear(
        input_features,
        len(class_names)
    )
)


# ============================================================
# Load trained weights
# ============================================================

checkpoint = torch.load(
    MODEL_PATH,
    map_location=DEVICE,
    weights_only=False
)


if "model_state_dict" in checkpoint:

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

else:

    model.load_state_dict(
        checkpoint
    )


model = model.to(
    DEVICE
)

model.eval()


# ============================================================
# Prediction function
# ============================================================

def predict_image(image_path):

    image = Image.open(
        image_path
    )

    image = ImageOps.exif_transpose(
        image
    )

    image = image.convert(
        "RGB"
    )

    tensor = transform(
        image
    )

    tensor = tensor.unsqueeze(
        0
    ).to(DEVICE)


    with torch.no_grad():

        output = model(
            tensor
        )

        probabilities = torch.softmax(
            output,
            dim=1
        )[0]


    predicted_index = int(
        torch.argmax(
            probabilities
        ).item()
    )


    prediction = class_names[
        predicted_index
    ]


    fake_probability = float(
        probabilities[0].item()
    )


    real_probability = float(
        probabilities[1].item()
    )


    confidence = float(
        probabilities[
            predicted_index
        ].item()
    )


    result = {

        "prediction":
            prediction,

        "confidence":
            confidence,

        "fake_probability":
            fake_probability,

        "real_probability":
            real_probability
    }


    return result


# ============================================================
# Command-line interface
# ============================================================

if __name__ == "__main__":

    parser = argparse.ArgumentParser()

    parser.add_argument(
        "--image",
        required=True,
        help="Path to image"
    )

    args = parser.parse_args()


    result = predict_image(
        args.image
    )


    print(
        f"Prediction: "
        f"{result['prediction']}"
    )

    print(
        f"Confidence: "
        f"{result['confidence'] * 100:.2f}%"
    )

    print(
        f"Fake Probability: "
        f"{result['fake_probability'] * 100:.2f}%"
    )

    print(
        f"Real Probability: "
        f"{result['real_probability'] * 100:.2f}%"
    )
'''


INFERENCE_PATH = (
    "/content/inference.py"
)


with open(
    INFERENCE_PATH,
    "w"
) as f:

    f.write(
        inference_code
    )


print(
    "Created:",
    INFERENCE_PATH
)
```
