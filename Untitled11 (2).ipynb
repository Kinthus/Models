```python
# !pip install -q transformers datasets torch scikit-learn accelerate matplotlib seaborn

import pandas as pd
import numpy as np
import torch
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import Dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    auc,
)
from transformers import (
    BertTokenizerFast,
    BertForSequenceClassification,
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback,
)
```

```python
import pandas as pd

file_path = "/content/news_dataset.csv"

df = pd.read_csv(file_path, low_memory=False)

print("Columns:")
print(df.columns.tolist())

print("\nShape:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nLabel values:")
print(df["label"].value_counts(dropna=False).head(20))
```

```python
# Convert label to numeric.
# Invalid values become NaN.
df["label"] = pd.to_numeric(df["label"], errors="coerce")

# Remove rows where label could not be converted
df = df.dropna(subset=["label"])

# Convert to integer
df["label"] = df["label"].astype(int)

# Keep only valid labels
df = df[df["label"].isin([0, 1])]

print("Final dataset size:", len(df))
print("\nLabel distribution:")
print(df["label"].value_counts())
```

```python
# Combine title + text
df["content"] = (
    df["title"].fillna("") + ". " + df["text"].fillna("")
).str.strip()

# Remove empty content
df = df[df["content"].str.len() > 0]

# Remove duplicates
df = df.drop_duplicates(subset=["content"])

print(f"\nFinal dataset size: {len(df)} rows")
print(df["label"].value_counts())
```

```python
train_texts, temp_texts, train_labels, temp_labels = train_test_split(
    df["content"].tolist(),
    df["label"].tolist(),
    test_size=0.2,
    random_state=42,
    stratify=df["label"],
)

val_texts, test_texts, val_labels, test_labels = train_test_split(
    temp_texts, temp_labels, test_size=0.5, random_state=42, stratify=temp_labels
)

print(f"Train: {len(train_texts)} | Val: {len(val_texts)} | Test: {len(test_texts)}")
```

```python
MODEL_NAME = "bert-base-uncased"
MAX_LEN = 256  # increase to 512 if you have GPU memory to spare

tokenizer = BertTokenizerFast.from_pretrained(MODEL_NAME)

def tokenize(texts):
    return tokenizer(
        texts,
        padding="max_length",
        truncation=True,
        max_length=MAX_LEN,
    )

train_encodings = tokenize(train_texts)
val_encodings = tokenize(val_texts)
test_encodings = tokenize(test_texts)


class NewsDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx])
        return item


train_dataset = NewsDataset(train_encodings, train_labels)
val_dataset = NewsDataset(val_encodings, val_labels)
test_dataset = NewsDataset(test_encodings, test_labels)
```

```python
model = BertForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)
```

```python
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    precision, recall, f1, _ = precision_recall_fscore_support(
        labels, preds, average="binary"
    )
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc, "precision": precision, "recall": recall, "f1": f1}
```

```python
training_args = TrainingArguments(
    output_dir="./bert-fake-news",
    num_train_epochs=3,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    learning_rate=2e-5,
    weight_decay=0.01,
    warmup_ratio=0.1,
    logging_dir="./logs",
    logging_steps=50,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    fp16=torch.cuda.is_available(),
    report_to="none",
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=2)],
)
```

```python
trainer.train()
```

```python
history = trainer.state.log_history

train_loss = [(h["step"], h["loss"]) for h in history if "loss" in h]
eval_loss = [(h["epoch"], h["eval_loss"]) for h in history if "eval_loss" in h]
eval_acc = [(h["epoch"], h["eval_accuracy"]) for h in history if "eval_accuracy" in h]
eval_f1 = [(h["epoch"], h["eval_f1"]) for h in history if "eval_f1" in h]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

if train_loss:
    steps, losses = zip(*train_loss)
    axes[0].plot(steps, losses, label="Train loss", color="tab:blue")
if eval_loss:
    epochs_l, losses_l = zip(*eval_loss)
    axes[0].plot(
        [e / max(epochs_l) * max(steps) for e in epochs_l] if train_loss else epochs_l,
        losses_l,
        label="Validation loss",
        color="tab:orange",
        marker="o",
    )
axes[0].set_xlabel("Training step")
axes[0].set_ylabel("Loss")
axes[0].set_title("Training vs Validation Loss")
axes[0].legend()

if eval_acc:
    epochs_a, accs = zip(*eval_acc)
    axes[1].plot(epochs_a, accs, label="Validation accuracy", marker="o")
if eval_f1:
    epochs_f, f1s = zip(*eval_f1)
    axes[1].plot(epochs_f, f1s, label="Validation F1", marker="s")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Score")
axes[1].set_title("Validation Accuracy & F1 over Epochs")
axes[1].legend()

plt.tight_layout()
plt.savefig("training_curves.png", dpi=150)
plt.show()
```

```python
test_results = trainer.evaluate(test_dataset)
print("Test set results:", test_results)

test_predictions = trainer.predict(test_dataset)
test_logits = test_predictions.predictions
test_probs = torch.softmax(torch.tensor(test_logits), dim=-1).numpy()
test_preds = np.argmax(test_logits, axis=-1)
test_true = test_predictions.label_ids
```

```python
cm = confusion_matrix(test_true, test_preds)

plt.figure(figsize=(5, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Real (0)", "Fake (1)"],
    yticklabels=["Real (0)", "Fake (1)"],
)
plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.title("Confusion Matrix — Test Set")
plt.tight_layout()
plt.savefig("confusion_matrix.png", dpi=150)
plt.show()
```

```python
fpr, tpr, _ = roc_curve(test_true, test_probs[:, 1])
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(5, 5))
plt.plot(fpr, tpr, color="tab:blue", label=f"ROC curve (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], color="gray", linestyle="--", label="Random guess")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve — Test Set")
plt.legend(loc="lower right")
plt.tight_layout()
plt.savefig("roc_curve.png", dpi=150)
plt.show()
```

```python
trainer.save_model("./bert-fake-news-final")
tokenizer.save_pretrained("./bert-fake-news-final")
print("Model saved to ./bert-fake-news-final")
```

```python
def predict(text: str):
    inputs = tokenizer(
        text, return_tensors="pt", truncation=True, padding=True, max_length=MAX_LEN
    ).to(model.device)
    with torch.no_grad():
        outputs = model(**inputs)
    probs = torch.softmax(outputs.logits, dim=-1)
    pred = torch.argmax(probs, dim=-1).item()
    return {"label": pred, "confidence": probs[0][pred].item()}

# Example:
# print(predict("Breaking: Scientists confirm shocking new discovery..."))
```
